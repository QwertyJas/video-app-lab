# video-app lab 01 - GPU facts
Runtime must be set to **T4 GPU** (Runtime > Change runtime type). Then Run all.


```python
import os, platform, shutil, subprocess, torch
print('python   :', platform.python_version())
print('torch    :', torch.__version__)
print('cuda ok  :', torch.cuda.is_available())
if torch.cuda.is_available():
    p = torch.cuda.get_device_properties(0)
    print('gpu      :', p.name)
    print('vram GB  :', round(p.total_memory/1024**3, 1))
    print('bf16 ok  :', torch.cuda.is_bf16_supported())
print('ram GB   :', round(os.sysconf('SC_PAGE_SIZE')*os.sysconf('SC_PHYS_PAGES')/1024**3, 1))
print('disk GB  :', round(shutil.disk_usage('/').free/1024**3, 1), 'free')
try:
    print(subprocess.run(['nvidia-smi','--query-gpu=name,memory.total','--format=csv'],
                         capture_output=True, text=True, timeout=15).stdout.strip())
except Exception as e:
    print('nvidia-smi not available:', type(e).__name__)
print()
print('VERDICT:', 'GPU READY' if torch.cuda.is_available() else 'CPU ONLY - switch Runtime to T4 GPU')
```

Expected: `gpu : Tesla T4`, `vram GB : 15.x`, `VERDICT: GPU READY`.
If it says CPU ONLY, the runtime type was not saved.
