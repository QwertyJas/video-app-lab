# video-app lab 05 - preflight guard (run cell 1 ONLY)
Cell 1 refuses to continue unless a GPU is actually attached, and prints RAM/VRAM/disk.
That way no model is ever downloaded into a CPU-only 12.7 GB runtime (the crash pattern we keep hitting).


```python
import os, shutil, psutil, torch
vm = psutil.virtual_memory()
print(f'RAM   : {vm.used/1024**3:.1f} used / {vm.total/1024**3:.1f} GB total')
print('disk  :', round(shutil.disk_usage('/').free/1024**3,1), 'GB free')
print('torch :', torch.__version__, '| cuda:', torch.cuda.is_available())
if not torch.cuda.is_available():
    raise SystemExit('STOP: this runtime has NO GPU. Runtime > Change runtime type > T4 GPU > Save, then reconnect.')
p = torch.cuda.get_device_properties(0)
print('GPU   :', p.name, '| VRAM', round(p.total_memory/1024**3,1), 'GB')
print('PREFLIGHT OK - safe to continue')
```
